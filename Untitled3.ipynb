```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

print("Libraries imported successfully!")

```

```python
# Languages  USE Hugging Face
# https://huggingface.co/datasets/papluca/language-identification
''' arabic (ar), bulgarian (bg), german (de), modern greek (el), english (en), spanish (es),
 french (fr), hindi (hi), italian (it), japanese (ja), dutch (nl), polish (pl), portuguese (pt),
  russian (ru), swahili (sw), thai (th), turkish (tr), urdu (ur), vietnamese (vi), and chinese (zh)
Dataset Structure'''
```

```python

!pip -q install datasets

from datasets import load_dataset

# Directly load dataset from Hugging Face
dataset = load_dataset("papluca/language-identification")

# Display available data splits
print(dataset)
```

```python
# convert into pandas
train_df = dataset["train"].to_pandas()

print("Dataset loaded successfully!")
print("Shape:", train_df.shape)
print(train_df.head())

print("\nLanguages:", train_df["labels"].nunique())
print(train_df["labels"].value_counts())
```

```python
# prepare datasets


# Check the dataset
print("Dataset shape:", train_df.shape)
print("Missing values:")
print(train_df.isnull().sum())

# Remove missing and duplicate records
train_df = train_df.dropna(subset=["text", "labels"])
train_df = train_df.drop_duplicates(subset=["text", "labels"])

# Keep only non-empty text
train_df["text"] = train_df["text"].astype(str).str.strip()
train_df = train_df[train_df["text"] != ""]

print("\nCleaned dataset shape:", train_df.shape)
print("\nLanguage distribution:")
print(train_df["labels"].value_counts())
```

```python
# TF-IDF Logistic regression


from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

X_train = train_df["text"]
y_train = train_df["labels"]

model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            analyzer="char",
            ngram_range=(2, 5),
            min_df=2,
            max_features=200000
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            random_state=42
        )
    )
])

model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
# model test

def detect_language(text):
    prediction = model.predict([text])[0]
    return prediction

print(detect_language("Hello, how are you?"))
print(detect_language("Bonjour, comment allez-vous?"))
print(detect_language("Hola, como estas?"))
print(detect_language("नमस्ते, आप कैसे हैं?"))
```

```python
# model evaluation

test_df = dataset["test"].to_pandas()

test_df = test_df.dropna(subset=["text", "labels"])
test_df = test_df.drop_duplicates(subset=["text", "labels"])

X_test = test_df["text"].astype(str)
y_test = test_df["labels"]

print("Test samples:", len(test_df))
print("Test languages:", y_test.nunique())
```

```python
# accuracy

from sklearn.metrics import accuracy_score

# Predict languages for test examples
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

```python
# classification report

from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
# confusion matrix

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    xticks_rotation=90,
    cmap="Blues",
    colorbar=False
)

plt.title("Language Detection Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# download file

import joblib

joblib.dump(model, "language_detection_model.pkl")

print("Model saved successfully!")
```
